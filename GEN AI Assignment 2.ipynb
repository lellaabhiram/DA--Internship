# Running Sum of 1D Array

```python
n=list(map(int,input().split())) # input is in list format
s=[] # empty list
b=0
for i in n:
    b+=i
    s.append(b) # we can only append the values into list
print(s)
```

# Shuffle the Array

```python
num=list(map(int,input().split())) # input in list format of even length
n=int(input()) # half of the input length
n1=num[0:n]
n2=num[n:]
n3=[]
for i in range(0,n):
    n3.append(n1[i])
    n3.append(n2[i])
print(n3)
```

# Kids With the Greatest Number of Candies

```python
candies=list(map(int,input().split()))
extraCandies=int(input())
a=max(candies)
c=[]
for i in range(len(candies)):
    if candies[i]+extraCandies>=a:
        c.append(True)
    else:
        c.append(False)
print(c)
```
